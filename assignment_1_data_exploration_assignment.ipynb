#Data Manipulation with Pandas:
Create a sample dataset using a dictionary.
Convert it into a Pandas DataFrame and display it.
Add a Markdown cell to describe your data.

```python
import pandas as pd
```

```python
#sample data set as dictionry
data = {
    'Name': ['Kelly', 'jenn', 'Alen'],
    'Age' : [ 25, 30, 34],
    'City': ['kitchner', 'waterloo', 'Cambridge']
}

#covert to dataframe
df = pd.DataFrame(data)

#display the dataframe
print(df)
```

## Sample Dataset

This dataset contains basic information about three people:
their name, age, and the city they live in. It is stored in a
Pandas DataFrame with 3 rows and 3 columns.

#Data Visualization with Matplotlib/Seaborn:
Create a bar chart or line plot to visualize your data.
Add a Markdown cell explaining the visualization.


```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Bar chart - age for each person
plt.bar(df['Name'], df['Age'])
plt.xlabel('Name')
plt.ylabel('Age')
plt.title('Age of People')
plt.show()
```

# Visualization

This bar chart shows the age of each person in the dataset.
The x-axis lists each person's name and the y-axis shows their
age, making it easy to compare ages at a glance.
