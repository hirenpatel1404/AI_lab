# Task 3: Handling Null Values

```python
# Create a DataFrame with a null value (np.nan) in each column
df_nulls = pd.DataFrame({
    "Name": ["Alice", np.nan, "Charlie"],
    "Age": [25, 30, np.nan],
    "City": [np.nan, "San Francisco", "Los Angeles"]
})
```

```python
# Show the table with nulls
print("Before filling nulls:")
print(df_nulls)
```

```python
# Check how many nulls are in each column
print("\nNull count per column:")
print(df_nulls.isnull().sum())
```

```python
# Fill nulls: 0 for the numeric Age column, "Unknown" for text columns
df_filled = df_nulls.fillna({"Name": "Unknown", "Age": 0, "City": "Unknown"})
```

```python
#REPLACE AGE 0 WITH MEAN VALUE
df_filled["Age"] = df_filled["Age"].replace(0, df_filled["Age"].mean())
print(df_filled)
```

```python
df_filled["Age"] = df_filled["Age"].replace(18.3333, round(df_filled["Age"].mean()))
df_filled["Age"] = df_filled["Age"].astype(int)
print(df_filled)
```

# Task 3 Interpretation
I made a table where each column has one missing value (`NaN`).
- `isnull().sum()` counts the missing values in each column.
- `fillna()` replaces them: I used **"Unknown"** for the text columns (Name and City), because filling a name or city with 0 would not make sense.
- For the numeric **Age** column, I first filled the missing value, then replaced it with the **mean (average) age**, which is a more realistic value than 0.

#Rounding the Age
- I used `round()` to turn the mean (e.g. 18.33) into a clean whole number, then `astype(int)` to remove the `.0` decimal.
- Note: `astype(int)` on its own does **not** round — it chops off the decimal (18.99 would become 18, not 19). That is why I round **first**, then convert the type.


# Task 6: Tuples and Sets

```python
# Create a tuple of fruits (tuples use round brackets and CANNOT be changed)
fruits = ("apple", "banana", "cherry")
print(fruits)
```

```python
# Create a set of numbers (sets use curly brackets and hold no duplicates)
numbers_set = {1, 2, 3, 4, 5}
print(numbers_set)
```

```python
# Try to add to the SET — this works, sets are changeable (mutable)
numbers_set.add(6)
print("\nSet after adding 6:", numbers_set)
```

```python
# Try to add to the TUPLE — this fails, tuples are unchangeable (immutable)
try:
    fruits.add("mango")
except AttributeError as e:
    print("\nCannot add to tuple:", e)
```

#Task 6 Interpretation
- A **set** is *mutable* (changeable), so `.add()` works and 6 was added.
- A **tuple** is *immutable* (fixed), so it has no `.add()` method and trying to change it causes an error.
- I used a `try/except` block to catch the error instead of letting the program crash.
Key difference: tuples are locked after creation; sets can grow and shrink. Sets also automatically remove duplicates.

#Task 8: Functions and Lambda & Task 10: Map, Reduce, and Filter

```python
# Regular function to square a number
def square(n):
    return n * n
```

```python
# Lambda function — a short one-line way to do the same thing
square_lambda = lambda n: n * n
```

```python
# Test both with two different inputs
print("Regular function:", square(4), square(7))
print("Lambda function:", square_lambda(4), square_lambda(7))
```

```python
#Lambda with two inputs:
# Add two numbers
add = lambda a, b: a + b
print(add(3, 7))

# Multiply two numbers
multiply = lambda a, b: a * b
print(multiply(4, 5))
```

```python
#Lambda with a condition (returns one of two things):
even_or_odd = lambda x: "even" if x % 2 == 0 else "odd"
print(even_or_odd(3))
print(even_or_odd(4))

```

```python
#Lambda on text:
# Make text uppercase
shout = lambda word: word.upper()
print(shout("I am doctor"))
```

```python
# Get the length of a word
length = lambda word: len(word)
print(length("banana"))
```

```python
#lambda in inside map and filter

nums = [1, 2, 3, 4, 5]

# Square every number
print(list(map(lambda x: x * x, nums)))
```

```python
# Keep only even numbers
print(list(filter(lambda x: x % 2 == 0, nums)))
```

### Task 8 Interpretation
Both functions do the same job — square a number.
- The **regular function** uses `def` and is good for longer logic.
- The **lambda** is a shortcut for simple one-line functions.
Both gave the same results (16 and 49), showing they are equivalent for this task.

**I also tried some other lambda examples** to understand them better — such as doubling a number, adding two numbers, checking even or odd, and using lambdas inside `map()` and `filter()`.

**Takeaway:** lambdas are best for short, simple, one-time functions. For anything longer or reused, a regular `def` function is clearer.

# Task 9: Iterators and Generators

```python
# ITERATOR: a class that produces the first 5 even numbers
class EvenNumbers:
    def __init__(self):
        self.current = 0      # start before the first even number
        self.count = 0        # how many we've given out

    def __iter__(self):
        return self           # the object is its own iterator

    def __next__(self):
        if self.count < 5:    # stop after 5 numbers
            self.current += 2
            self.count += 1
            return self.current
        else:
            raise StopIteration   # signal that we're done

print("Iterator result:")
for num in EvenNumbers():
    print(num, end=" ")
```

```python
# GENERATOR: a function that does the same thing, much shorter
def even_generator():
    num = 2
    for _ in range(5):
        yield num          # 'yield' hands back one value at a time
        num += 2

print("\n\nGenerator result:")
for num in even_generator():
    print(num, end=" ")
```

# Task 9 Interpretation
Both produce the first 5 even numbers (2 4 6 8 10).
- The **iterator** is a class using `__iter__` and `__next__`. It works but takes more code.
- The **generator** is a function using `yield`. It does the same thing in far fewer lines.
`yield` gives back one value each time it's asked, which makes generators a simpler way to build iterators.

# Task 11: Object-Oriented Programming - Creating a Class

```python
# A class is a blueprint for creating objects
class Rectangle:
    # __init__ runs when a new Rectangle is created; it sets the attributes
    def __init__(self, length, width):
        self.length = length
        self.width = width

    # Method to calculate area
    def area(self):
        return self.length * self.width

    # Method to calculate perimeter
    def perimeter(self):
        return 2 * (self.length + self.width)

# Create two Rectangle objects
rect1 = Rectangle(5, 3)
rect2 = Rectangle(10, 4)

print("Rectangle 1 → Area:", rect1.area(), "Perimeter:", rect1.perimeter())
print("Rectangle 2 → Area:", rect2.area(), "Perimeter:", rect2.perimeter())
```

### Task 11 Interpretation
- The **Rectangle** class is a blueprint. `length` and `width` are its **attributes** (data), and `area()` and `perimeter()` are its **methods** (actions).
- `__init__` sets up each object's attributes when it's created.
- I made two objects with different sizes, and each used the same methods to give its own results — this shows how one class can create many objects.


# Task 12: Pandas Data Analysis

```python
df_employees = pd.DataFrame({
    'Name': ['John', 'Jane', 'Bob', 'Alice', 'Charlie'],
    'Department': ['IT', 'HR', 'IT', 'Finance', 'HR'],
    'Salary': [55000, 65000, 70000, 60000, 58000]
})
print(df_employees)
```

```python
# 1. Average salary by department —
# groupby() groups rows by Department,
# we take the mean of Salary for each group
average_salary_by_dept = df_employees.groupby('Department')['Salary'].mean()
print(average_salary_by_dept)
```

```python
#Get the names of employees with salary > 60000
high_salary_employees = df_employees[df_employees['Salary'] > 60000][['Name', 'Salary']]
print(high_salary_employees)
```

```python
#Add a new column 'Bonus' which is 10% of salary
df_employees['Bonus'] = df_employees['Salary'] * 0.10
df_employees['Bonus'] = df_employees['Bonus'].astype(int)
print(df_employees)
```

```python
#update salary after adding bonus
df_employees['Salary_updated'] = df_employees['Salary'] + df_employees['Bonus']
print(df_employees)
```

### Task 12 Interpretation (Bonus and Updated Salary)
- **Bonus column:** I created a new column by multiplying Salary by 0.10 (10%). Pandas applied this to every row automatically.
- **Salary_updated column:** I then added Salary + Bonus together to show each employee's total pay after the bonus.
- This gives two new columns in the output: the bonus amount on its own, and the final salary including the bonus — so it's easy to see both the extra and the new total side by side.
