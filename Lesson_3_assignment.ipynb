# 2. Lists

```python
books = ["Python Basics", "Advanced Python", "Data Science with Python"]
books.append("Machine Learning")
books.insert(1, "Web Development")
print("Updated book list:", books)
```

### Part 2 Interpretation: Lists
A list holds multiple items in order. I added two books using two different methods:

# 5. Functions and Conditional Statements

```python
def check_availability(book_name):
    # Check if the book is in the list using the "in" keyword
    if book_name in books:
        return True
    else:
        return False

# Test the function
print(f"Is 'Python Basics' available? : {check_availability('Python Basics')}")
print(f"Is 'Harry Potter' available? {check_availability('Harry Potter')}")
```

### Part 5 Interpretation: Functions and Conditional Statements
I completed a function that checks whether a book exists in the list.
- The **`in`** keyword checks if `book_name` is present in the `books` list.
- If it is found, the function returns **`True`** (available).
- If not, it returns **`False`** (not available).

This combines a **function** (reusable block of code) with an **`if/else` conditional** (making a decision based on a condition).

# 6. File I/O

```python
def save_books_to_file(filename):
    with open(filename, 'w') as f:      # 'w' = write mode
        for book in books:
            f.write(f"{book}\n")        # \n puts each book on a new line
```

```python
# Read books back from the file
def read_books_from_file(filename):
    with open(filename, 'r') as f:      # 'r' = read mode
        content = f.read()              # read everything in the file
    return content
```

```python
# Test file I/O
save_books_to_file("books.txt")
!cat books.txt  # Display file contents in Colab

```

```python
# Test the read function
print("\nBooks read from file:")
print(read_books_from_file("books.txt"))
```

### Part 6 Interpretation: File I/O
- **Save function** writes each book to a text file, one per line (`'w'` = write mode).
- **Read function** opens the same file (`'r'` = read mode) and reads all the text back.
- `with open(...)` safely opens the file and closes it automatically when done.

This shows how a program can save data to a file and load it back later.

# 7. Pickle for serialization

```python
def save_books_pickle(filename):
    with open(filename, 'wb') as f:     # 'wb' = write binary
        pickle.dump(books, f)           # dump = save the list into the file

# Load books back from the pickle file
def load_books_pickle(filename):
    with open(filename, 'rb') as f:     # 'rb' = read binary
        loaded_books = pickle.load(f)   # load = read the list back
    return loaded_books

# Test pickle functions
save_books_pickle("books.pkl")
print("Books loaded from pickle:", load_books_pickle("books.pkl"))
```

### Part 7 Interpretation: Pickle
Pickle saves a whole Python object (my books list) to a file and loads it back later.
- **`pickle.dump()`** saves the list into the file (`'wb'` = write binary).
- **`pickle.load()`** reads the list back out (`'rb'` = read binary).

Unlike a text file, pickle keeps the data as a real Python list, so it comes back ready to use.

# 8. SQLite Database Interaction

```python

def create_books_table():
    conn = sqlite3.connect('library.db')
    cursor = conn.cursor()
    cursor.execute('''CREATE TABLE IF NOT EXISTS books
                      (id INTEGER PRIMARY KEY, title TEXT, quantity INTEGER)''')
    conn.commit()
    conn.close()
```

```python
# Add a book to the database
def add_book_to_db(title, quantity):
    conn = sqlite3.connect('library.db')     # open connection to the database
    cursor = conn.cursor()                   # cursor runs the commands
    # Insert a new row. The ? marks are placeholders filled by the values safely
    cursor.execute("INSERT INTO books (title, quantity) VALUES (?, ?)", (title, quantity))
    conn.commit()                            # save the change
    conn.close()                             # close the connection
```

```python
# Test database functions
create_books_table()
add_book_to_db("Python for Beginners", 10)
```

```python
# Read the table back to confirm it worked
conn = sqlite3.connect('library.db')
cursor = conn.cursor()
cursor.execute("SELECT * FROM books")
print("Books in database:", cursor.fetchall())
conn.close()
```

### Part 8 Interpretation: SQLite Database
SQLite stores data in a database file (`library.db`).
- **`connect()`** opens the database, and a **cursor** runs the commands.
- **`INSERT INTO`** adds a new book row with a title and quantity.
- The **`?` placeholders** safely fill in the values (this prevents errors and security issues).
- **`commit()`** saves the change, and **`close()`** ends the connection.

I also read the table back with **`SELECT *`** to confirm the book was added.

# 9. Web API Interaction (GitHub API)

```python
def get_python_repos():
    url = "https://api.github.com/search/repositories"   # GitHub's search address
    # params = the search settings: Python language, sorted by most stars
    params = {"q": "language:python", "sort": "stars", "order": "desc"}
    response = requests.get(url, params=params)          # send the request
    data = response.json()                               # convert the reply to a Python dict
    # Take the first 5 results and pull out just each repo's name
    return [repo['name'] for repo in data['items'][:5]]

# Test API function
print("Top 5 Python repositories on GitHub:", get_python_repos())

print("\nLibrary Management System operations completed.")
```

### Part 9 Interpretation: Web API Interaction
This part fetches live data from the internet using the GitHub API.
- **`requests.get()`** sends a request to GitHub's search address with settings (`params`) asking for Python repos sorted by most stars.
- **`.json()`** turns GitHub's reply into a Python dictionary I can read.
- A **list comprehension** pulls out just the names of the top 5 repos.

This shows how a program can pull real, up-to-date data from an outside service instead of only using data stored locally.
